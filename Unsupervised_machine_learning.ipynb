```python
data = data.drop('quality',axis =1) #unsupervised machine learning work on unlabeled dataset
```

```python
data.head(2)
```

```python
# PCA (Principle component analysis )
# concerting Nd data into 2d or 3d such that do not less their co-relation or co-structure
#Dimentional Reduction
#
#Why PCA used
#pca use for data overall structure
#see outliers
#see the cluster given by unsupervised machine learning

#Scaling : without changing correlation put in similar categoris
#1> min max scaler : either 0 to 1 ko bich ma data lai rakhxa
# #2> StandardScalar : convert the values such that c0/1 standard deviation (std)=1
# c0/2 std = 1
```

```python
#standard Scaler
#import scaler
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
#scaled data
data_scaled = scaler.fit_transform(data)
```

```python
data_scaled
```

```python
#convert data into dataframe
data_scaled = pd.DataFrame(data_scaled,columns = data.columns)
```

```python
data_scaled
```

```python
#Data decompostion using PCA
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
```

```python
pca_data = pca.fit_transform(data)
```

```python
pca_data
```

```python
pca_data = pd.DataFrame(pca_data,columns = ['pca1','pca2'])
```

```python
pca_data.head(3)
```

```python
#plot the data pca_data
plt.figure(figsize=(10,6))
sns.scatterplot(x = 'pca1',y = 'pca2',data = pca_data)
plt.title('wine data befor scaled')
plt.xlabel('PCA1')
plt.ylabel('PCA2')
plt.show()
```

```python
#
pca_scaled = pca.fit_transform(data_scaled)
```

```python
pca_scaled = pd.DataFrame(pca_scaled,columns = ['pca_1','pca_2'])
```

```python
#plot
plt.figure(figsize=(10,6))
sns.scatterplot(x = 'pca_1',y = 'pca_2',data = pca_scaled)
plt.title('wine data after scaled')
plt.xlabel('PCA1')
plt.ylabel('PCA2')
```

```python
#visulizing both
fig,axes = plt.subplots(1,2,figsize=(10,5))
axes[0].set_title("Befor Scaled")
sns.scatterplot(x = 'pca1',y = 'pca2',data = pca_data,label = 'before scaled',ax = axes[0],color ='red')
axes[1].set_title('After Scaled')
sns.scatterplot(x='pca_1',y = 'pca_2',data = pca_scaled,label ='after scaled',ax = axes[1])
plt.show()
```

```python
#KMeans clusteing
from sklearn.cluster import KMeans
```

```python
wcss = []
for x in range(2,9):
  k = KMeans(n_clusters=x)
  k.fit_transform(pca_scaled)
  wcss.append(k.inertia_)
```

```python
wcss
```

```python
#plot the wcss with respective clusetesrs
plt.figure(figsize = (10,6))
plt.title('wcss values for every possible clusters')
plt.plot(range(2,9),wcss,marker='D')
plt.xlabel('number of clusters')
plt.ylabel('Wcss')
plt.show()
```

```python
#since from the observation of the above diagram the optimal number of clusters is 5
```

```python
#new model
kmeans = KMeans(n_clusters=5)
```

```python
data_scaled['clusters'] = kmeans.fit_predict(data_scaled)
data_scaled['PCA1'] = pca_scaled['pca_1']
data_scaled['PCA2'] = pca_scaled['pca_2']
```

```python
#plot
sns.scatterplot(x='PCA1',y ='PCA2',hue = 'clusters',data = data_scaled,palette='bright')
```
