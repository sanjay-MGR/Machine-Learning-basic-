```python
#import laiberies
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.metrics import accuracy_score,r2_score
from sklearn.preprocessing import StandardScaler
```

```python
#load datasets
data = sns.load_dataset('iris')
```

```python
data.head(4)
```

```python
data.isna().sum()
```

```python
#with out scaled
data1 = data.copy()
```

```python
a = data1.drop('species',axis=1)
```

```python
b = data1['species']
```

```python
a_train,a_test,b_train,b_test = train_test_split(a,b,test_size = 0.2,random_state = 42)
```

```python
#build the model
kNN_eu = KNeighborsClassifier(n_neighbors=5,metric = 'euclidean')
kNN_eu
```

```python
kNN_man = KNeighborsClassifier(n_neighbors=5,metric = 'manhattan')
kNN_man
```

```python
#train the model
kNN_eu.fit(a_train,b_train)
```

```python
kNN_man.fit(a_train,b_train)
```

```python
#predic the value
b_pred_traineu = kNN_eu.predict(a_train)
b_pred_traineu
```

```python
b_pred_trainman = kNN_man.predict(a_train)
b_pred_trainman
```

```python
b_test_eu = kNN_eu.predict(a_test)
b_test_eu
```

```python
b_test_man = kNN_man.predict(a_test)
b_test_man
```

```python
#check the accuracy
acs_train_euu = accuracy_score(b_train,b_pred_traineu)
acs_train_euu
```

```python
acs_train_mann = accuracy_score(b_train,b_pred_trainman)
acs_train_mann
```

```python
acs_test_euu = accuracy_score(b_test,b_test_eu)
acs_test_euu
```

```python
acs_test_mann = accuracy_score(b_test,b_test_man)
acs_test_mann
```
