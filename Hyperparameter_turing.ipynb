```python
#import laiberies
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#loaddataset
data = sns.load_dataset('tips')
```

```python
data.isna().sum()
```

```python
data.head(2)
```

```python
#encode
enc_sex = ({'Male':1,'Female':0})
enc_smoker = ({'Yes':1,'No':0})
enc_time = ({'Lunch':0,'Dinner':1})
enc_day =({'Thur':0,'Fri':1,'Sat':2,'Sun':3})
```

```python
data['sex']=data['sex'].map(enc_sex)
data['smoker']=data['smoker'].map(enc_smoker)
data['time']=data['time'].map(enc_time)
data['day']=data['day'].map(enc_day)
```

```python
data.head(2)
```

```python
#split data into x and y
x = data.drop('day',axis = 1)
y = data['day']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2, random_state = 42 )
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
#model build
model = DecisionTreeRegressor()
model
```

```python
#fit the model
model.fit(x_train,y_train)
```

```python
#predict the value
y_pred = model.predict(x_train)
y_pred
```

```python
y_pred_test = model.predict(x_test)
y_pred_test
```

```python
#check accuracy
from sklearn.metrics import r2_score
```

```python
r2_train = r2_score(y_train,y_pred)
r2_train
```

```python
r2_test = r2_score(y_test,y_pred_test)
r2_test
```

```python
# improve the model by puring the tree using GridSearchCV
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    'max_depth':[2,3,4,5,6,7,8,9],
    'min_samples_split':[2,4,10,12,16,18,24,28,38,40,45,50],
    'min_samples_leaf':[1,2,3,46,7,9,10,14,16,18,19],
    'max_leaf_nodes':[5,10,15,20,25,30,35,40,45]
}
```

```python
grid = GridSearchCV(estimator = model,param_grid = param_grid,cv = 5)
```

```python
#train the. model
grid.fit(x_train,y_train)
```

```python
print(f'best parameters are =  {grid.best_params_}')
```

```python
# build new model
model_new = DecisionTreeRegressor(max_depth = 4,max_leaf_nodes =10,min_samples_leaf=16,min_samples_split=45)
```

```python
model_new
```

```python
#train the model
model_new.fit(x_train,y_train)
```

```python
#predict the value
y_pred_new = model_new.predict(x_train)
y_pred_new
```

```python
y_pred_test_new = model_new.predict(x_test)
y_pred_test_new
```

```python
#check the accuracy
r2_train_new = r2_score(y_train,y_pred_new)
r2_train_new
```

```python
r2_test_new = r2_score(y_test,y_pred_test_new)
r2_test_new
```

```python
#using randomsearchcv
from sklearn.model_selection import RandomizedSearchCV
```

```python
dict_param = {
    'max_depth':[2,3,4,5,6,7,8,9],
    'min_samples_split':[2,4,10,12,16,18,24,28,38,40,45,50],
    'min_samples_leaf':[1,2,3,46,7,9,10,14,16,18,19],
    'max_leaf_nodes':[5,10,15,20,25,30,35,40,45]
}
```

```python
random = RandomizedSearchCV(estimator = model,param_distributions = dict_param,cv = 5)
```

```python
#fit the model
random.fit(x_train,y_train)
```

```python
#find best parameter
print(f'best parameters are = {random.best_params_}')
```

```python
#create new model with best paraameters
new_model = DecisionTreeRegressor(max_depth = 8,max_leaf_nodes = 35,min_samples_leaf = 16,min_samples_split= 45)
```

```python
#fit the model
new_model.fit(x_train,y_train)
```

```python
#predict the value
y_pred_new_new = new_model.predict(x_train)
y_pred_new_new
```

```python
y_pred_test_new_new = new_model.predict(x_test)
y_pred_test_new_new
```
