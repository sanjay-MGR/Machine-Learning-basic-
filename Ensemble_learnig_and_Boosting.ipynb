```python
##import laiberies
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#load dataset
data = sns.load_dataset('tips')
```

```python
data.head(3)
df = data.copy()
df.head(2)
```

```python
data = data[['total_bill','tip','day']]
```

```python
#split data into x and y
x = data.drop('day',axis = 1)
```

```python
y = data.day
```

```python
#import train_test_split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
```

```python
#import model
from sklearn.tree import DecisionTreeClassifier
```

```python
#build the model
base_model  = DecisionTreeClassifier()
base_model
```

```python
#fit the model
base_model.fit(x_train,y_train)
```

```python
#Boosting (AdaBoost)
```

```python
from sklearn.ensemble import AdaBoostClassifier
```

```python
ada = AdaBoostClassifier(estimator=base_model,n_estimators=100,random_state=42)
#n_estimators=100 it retrain the model 100 times
```

```python
#fit the model
ada.fit(x_train,y_train)
```

```python
#perdic the value
y_pred = ada.predict(x_test)
```

```python
#check accuracy
from sklearn.metrics import accuracy_score
```

```python
acs = accuracy_score(y_test,y_pred)
acs #accuracy is 32% deu to data altered
```

```python
#Regression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import AdaBoostRegressor
```

```python
#load dataset
data = sns.load_dataset('mpg')
```

```python
data.head(2)
```

```python
data1 = data[['mpg','cylinders','displacement','horsepower']]
```

```python
X = data1.drop('mpg',axis = 1)
X.isna().sum()
```

```python
X['horsepower'] = X['horsepower'].fillna(X['horsepower'].mean())
```

```python
X.isna().sum()
```

```python
Y = data1.mpg
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size = 0.2,random_state = 42)
```

```python
#build the base model
base_modelR = DecisionTreeRegressor()
base_modelR
```

```python
#fit the model
base_modelR.fit(X_train,Y_train)
```

```python
#build model
adaR = AdaBoostRegressor(estimator=base_modelR,n_estimators=100,random_state=42)
```

```python
#trian model
adaR.fit(X_train,Y_train)
```

```python
#predict the value
Y_pred = adaR.predict(X_test)
```

```python
#check accuracy
from sklearn.metrics import r2_score
r2_score = r2_score(Y_test,Y_pred)
r2_score
```
