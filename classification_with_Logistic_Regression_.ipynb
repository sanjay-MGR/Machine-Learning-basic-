```python
#mport laiberies
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,f1_score,precision_score,recall_score,confusion_matrix
```

```python
#load datasets
data = sns.load_dataset('titanic')
data.head(2)
```

```python
df1 = data[['survived','age','fare']]
```

```python
df1.head(2)
```

```python
df1.isna().sum() #check null value
```

```python
df1.shape
```

```python
df1['age'] = df1['age'].fillna(df1['age'].mean()) #fill the null value of age with its mean or average age
```

```python
df1.isna().sum()
```

```python
df1
```

```python
#split data into x and y
x = df1.drop('survived',axis = 1)
```

```python
x.head(2)
```

```python
y = df1['survived']
y.head(4)
```

```python
#model built
LR = LogisticRegression()
LR
```

```python
#fit the model
LR.fit(x,y)
```

```python
#predict the value
y_pred = LR.predict(x)
y_pred
```

```python
#check the accuracy,recall,precision,cnfucion matrix ,recall
acs = accuracy_score(y,y_pred)
acs
```

```python
#recall
reacall = recall_score(y,y_pred)
round(reacall,2)
```

```python
#percision
pas = precision_score(y,y_pred)
pas
```

```python
#f1_score
f1 = f1_score(y,y_pred)
f1
```

```python
#confuion matrix
cm = confusion_matrix(y,y_pred)#[tp,fn,fp,tn]
cm
```

```python
# multiple Classification with Logistic Regression
#Load data set
data_new = sns.load_dataset('penguins')
```

```python
data_new
```

```python
data_new = data_new.drop(['island','sex'],axis = 1)
```

```python
data_new.isna().sum()
```

```python
#handel or fill the null values with there mean
data_new['bill_length_mm'] = data_new['bill_length_mm'].fillna(data_new['bill_length_mm'].mean())
data_new['bill_depth_mm'] = data_new['bill_depth_mm'].fillna(data_new['bill_depth_mm'].mean())
data_new['flipper_length_mm'] = data_new['flipper_length_mm'].fillna(data_new['flipper_length_mm'].mean())
data_new['body_mass_g'] = data_new['body_mass_g'].fillna(data_new['body_mass_g'].mean())
```

```python
data_new.isna().sum()
```

```python
#split data into a and b
a = data_new.drop('species',axis = 1)
a.head(2)
```

```python
b = data_new['species']
b.head(2)
```

```python
#build the model
multiple_lr = LogisticRegression(max_iter=1000)
multiple_lr
```

```python
# train or fit the model
multiple_lr.fit(a,b)
```

```python
# predict the value
y_predrr =multiple_lr.predict(a)
y_predrr
```

```python
output = pd.DataFrame(y_predr,columns = ['Output'])
```

```python
output.head(10)
```

```python
# check accuracy
accuracy_sc  = accuracy_score(b,y_predrr)
accuracy_sc
```

```python
#confucion matrix
conf_mat = confusion_matrix(b,y_predrr)
conf_mat
```
